## Project Overview

This project analyzes 14 days of sales and stock records for Wanjiku's fresh-produce stall at Gikomba Market. The objective is to evaluate business performance, compare weekly sales, identify strong and weak trading days, assess stock levels, and provide data-driven recommendations to improve profitability and inventory management.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2


# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Sales Summary

I used Python's built-in functions to calculate total sales, average daily sales, the highest sales amount, and the lowest sales amount over the 14-day period. I also used list indexing to identify the corresponding day and week.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

best_day = days[highest_index]
worst_day = days[lowest_index]

best_week = (highest_index // 7) + 1
worst_week = (lowest_index // 7) + 1

print("Total Sales:", total_sales)
print("Average Daily Sales:", round(average_sales, 2))
print("Highest Sales:", highest_sales)
print("Best Day:", best_day, "| Week:", best_week)
print("Lowest Sales:", lowest_sales)
print("Worst Day:", worst_day, "| Week:", worst_week)
```

## Task 2: Weekly Performance

I divided the 14 daily sales figures into two groups using list slicing. I then calculated the total sales for each week and the percentage growth or decline from Week 1 to Week 2.

```python
def week_totals(sales):
    """Return the total sales for Week 1 and Week 2."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])

    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth = ((week2_total - week1_total) / week1_total) * 100

print("Week 1 Sales:", week1_total)
print("Week 2 Sales:", week2_total)
print("Week-on-Week Growth:", round(growth, 2), "%")
```

## Task 3: Daily Sales Classification

I created a function to classify daily sales based on the overall average. Days earning more than 20% above average are classified as Strong, while days earning more than 20% below average are classified as Weak. The remaining days are classified as Normal.

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal, or Weak."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


classification_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

daily_classifications = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)
    classification_counts[category] += 1

print("Daily Classifications:")

for day, amount, category in zip(days, daily_sales, daily_classifications):
    print(f"{day}: KES {amount:,} - {category}")

print("\nClassification Summary:")

for category, count in classification_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday vs Weekend Performance

I separated the sales into weekdays (Monday to Friday) and weekends (Saturday and Sunday). I calculated the average for each group and compared their performance using the percentage difference relative to weekday sales.

```python
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("Weekday Average:", round(weekday_average, 2))
print("Weekend Average:", round(weekend_average, 2))
print("Weekend Difference:", round(percentage_difference, 2), "%")
```

## Task 5: Stock and Gross Margin Analysis

I used a function to identify products whose available stock is at or below their reorder levels. I also calculated the gross margin percentage for each product to understand how much of the selling price remains after deducting the cost.

```python
def products_to_reorder(stock):
    """Return products with quantities at or below reorder levels."""
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_products = products_to_reorder(stock)

print("Products Requiring Restocking:")

for product in reorder_products:
    print(product)


print("\nGross Margin Analysis:")

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    print(f"{product}: {margin:.2f}%")
```

```python
def classify_day(amount, average):
    """Classify daily sales as Strong, Normal, or Weak."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


classification_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

daily_classifications = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_classifications.append(category)
    classification_counts[category] += 1

print("Classification Summary:")

for category, count in classification_counts.items():
    print(f"{category}: {count} days")
```
